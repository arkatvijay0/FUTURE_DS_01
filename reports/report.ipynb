# 1. Project Overview

The Business Sales Analysis project was developed to analyze an online retail dataset and generate meaningful business insights using Python and Microsoft Power BI.

The primary objective was to clean the raw sales data, perform exploratory data analysis (EDA), identify sales patterns, and develop an interactive dashboard that helps business stakeholders monitor key performance indicators such as revenue, customers, products, and country-wise sales.

The project demonstrates the complete Data Analytics workflow from raw data preprocessing to business visualization and reporting.

---

# 2. Problem Statement

Businesses generate thousands of sales transactions every day. Raw transactional data alone does not provide meaningful information for decision-making.

The objective of this project is to:

* Analyze business performance
* Track monthly revenue
* Identify best-selling products
* Identify highest revenue-generating customers
* Compare sales across countries
* Build an interactive dashboard for management

# 3. Objectives

# 5. Data Preprocessing

The following preprocessing steps were performed:

### Missing Value Handling

* Removed rows with missing Customer IDs
* Removed null values

### Duplicate Removal

* Removed duplicate transactions

### Invalid Data Handling

* Removed cancelled invoices
* Removed negative quantity values
* Removed negative unit prices

### Feature Engineering

Created:

* Sales = Quantity × Unit Price
* Month
* Month Number

These features improved analysis and dashboard visualization.

# 8. Key Business Insights

The analysis revealed several important business findings:

### Revenue Analysis

* Total Revenue exceeded 4 Million.
* Revenue fluctuated across months.

### Country Analysis

* United Kingdom contributed the majority of total revenue.
* Netherlands and Germany followed with significantly lower contributions.

### Customer Analysis

* A small number of customers generated a significant portion of revenue.

### Product Analysis

* Regency Cake Stand and Medium Ceramic Top Storage Jar were among the highest-selling products.
* Some premium products had high prices but lower sales volume.

### Sales Trend

* Monthly revenue showed seasonal fluctuations.
