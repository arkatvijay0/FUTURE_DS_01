```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"], dayfirst=True)
```

```python
df["InvoiceDate"].head()
```

```python
df.info()
```

```python
df["Month"] = df["InvoiceDate"].dt.month_name()
```

```python
df[["InvoiceDate", "Month"]].head(5)
```

```python
#Calculate Monthly Revenue
monthly_sales = (
    df.groupby("Month")["Sales"]
      .sum()
)

monthly_sales
```

```python
#Arrange Months in Calendar Order
month_order = [
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]

monthly_sales = monthly_sales.reindex(month_order)
```

```python
#Draw the Line Chart
plt.figure(figsize=(12,6))

monthly_sales.plot(
    kind="line",
    marker="o",
    linewidth=3,
    color="blue"
)

plt.title("Monthly Sales Trend", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.grid(True)

plt.tight_layout()

plt.savefig(r"C:\Users\arkat\OneDrive\Desktop\M.tech\b.techprojects\Internship projects\Business Sales Performance Analytics/monthly_sales_trend.png", dpi=300)

plt.show()
```
